#### Lowercasing:

````python
#pre-check:
print("Sample text before lowercasing:")
print(figure_df['Sentence'][:1])

#lowercase all the articles:
sentences_lower_cased = [art.lower() for art in figure_df['Sentence']]

#post-check:
print("Sample text after lowercasing:")
print(sentences_lower_cased[:1])
````

#### Tokenization:

````python
#TOKENIZATION, SIMPLE - 

#pre-check:
print("Sample text before tokenization:")
print(figure_df['Sentence'][:1])

#basic solution, using the string method `.split()`. 
sentences_tokenized = [art.split() for art in figure_df['Sentence']]

#post-check:
print("Sample text after tokenization:")
print(sentences_tokenized[:1])
````

````python
#TOKENIZATION, ADVANCED -

#more sophisticated solution, using the NLTK library.
#NOTE: TbWT separtes punctuation (world!), handles contractions (can't), and splits critics (it's).
from nltk.tokenize import TreebankWordTokenizer #for tokenization

#pre-check:
print("Sample text before advanced tokenization:")
print(trunk_df['Sentence'][:1])

#tokenize the articles:
sentences_tokenized = [TreebankWordTokenizer().tokenize(sentence) for sentence in trunk_df['Sentence']]

#post-check:
print("Sample text after advanced tokenization:")
print(sentences_tokenized[:1])
````

#### Stopwords:

````python
#STOPWORDS, SIMPLE - 

#let's use NLTK to get a list of stopwords:
from nltk.corpus import stopwords
#nltk.download("stopwords") #NOTE: uncomment this if needed. 

#let's create a list of stopwords:
mystopwords = stopwords.words("english")
mystopwords.extend(["add", "more", "words"]) # it's just a list, so we can add more words if we want to.

print(mystopwords) #just to demonstrate the point...
````

````python
# now, remove stopwords from the corpus:

#pre-check:
print("Sample text before removing stopwords:")
print(figure_df['Sentence'][:1])

#stopword removal:
sentences_without_stopwords = [] #initialize an empty list
for sentence in figure_df['Sentence']: #for each sentence
    sentence_no_stop = "" #initialize an empty string
    for word in sentence.lower().split(): #for each word in the list of lowered words from the sentence
        if word not in mystopwords: #if the word is not a stopword
            sentence_no_stop = sentence_no_stop + " " + word #add the word to the empty string
    sentences_without_stopwords.append(sentence_no_stop) #add the sentence sans stopwords to the list

#post-check:
print("Sample text after removing stopwords:")
print(sentences_without_stopwords[:2])
````

````python
# Optional: same solution, but with list comprehension:
# NOTE: can you explain how this works? 
sentences_without_stopwords = [" ".join([w for w in sentence.lower().split() if w not in mystopwords]) for sentence in figure_df['Sentence']]
print(sentences_without_stopwords[:1])
````

#### Stemming and lemmatization:

````python
#STEMMING AND LEMMATIZATION -

from nltk.stem.snowball import SnowballStemmer #this stems by removing suffixes (e.g. -ing, -ed)

stemmer = SnowballStemmer("english") #initialize the stemmer

#pre-check:
print("Sample text before stemming:")
print(figure_df['Sentence'][:1])

stemmed_text = [] #initialize an empty list
for sentence in figure_df['Sentence']: #for each sentence
    stemmed_words = "" #initialize an empty string
    for word in sentence.lower().split(): #for each word in the list of lowered words from the sentence
        stemmed_words = stemmed_words + " " + stemmer.stem(word) #stem the word and add it to the empty string
    stemmed_text.append(stemmed_words.strip()) #add the stemmed sentence to the list

#post-check:
print("Sample text after stemming:")
print(stemmed_text[:2])
````

````python
# compare tokeninzation and lemmatization using `Spacy`:

import spacy 
#spacy.cli.download("en_core_web_sm") #uncomment this if needed.
nlp = spacy.load("en_core_web_sm") #load the small english model.

#pre-check:
print("Sample text before lemmatization:")
print(figure_df['Sentence'][:1])

#let's lemmatize the articles:
lemmatized_sentences = [[token.lemma_ for token in nlp(sentence)] for sentence in figure_df['Sentence']]

#post-check:
print("Sample text after lemmatization:")
print(lemmatized_sentences[:1])
````

### 3. The machine can't read words: Turn them to numbers - Vectorization

<img src="../../images/03_vectorization.png" alt="Preprocessing diagram" style="max-width: 150px;">

Start with the basics: Counting words with CountVectorizer

````python
#COUNT VECTORIZER
from sklearn.feature_extraction.text import CountVectorizer #a vectorizer (converts text to numbers)



# Let's use a simple CountVectorizer without any fancy option as a baseline to inspect the effect preprocessing and tokenization can have on the document-term matrix:

# CountVectorizer needs to know how to tokenize the text. Since we don't want it's default preprocessing steps (see below for some optional parameters), 
# We can provide a custom tokenizer function to the CountVectorizer, to just get the raw words (without lowercasing, stopword removal, lemmatization, etc.).

def my_tokenizer(text):
    return text.split()  # just split the text into words based on whitespace. This is a simple tokenizer that does not perform any additional preprocessing.

# Let's feed this set of words per sentence to the CountVectorizer, and see what it does.

vectorizer_CV_raw = CountVectorizer(tokenizer=my_tokenizer, lowercase=False, token_pattern=None) # Note: we start with the original text for comparison.

# Now we can use this vectorizer to convert the text to a document-term matrix (DTM). 
# The DTM is a matrix where each row represents a document (here: a sentence) and each column represents a term (here: a word). 
# The values in the matrix represent the frequency of the term in the document.

# Fit the vectorizer to the text, and transform the text to a document-term matrix
# Let's just use two sentences for this example, to keep the output manageable.
# First print the sentences we are using for this example:
print("Sentences used for this example:")
print(figure_df['Sentence'][2:4].to_list())

# Now fit the vectorizer to the text, and transform the text to a document-term matrix.
X_CV = vectorizer_CV_raw.fit_transform(figure_df['Sentence'][2:4].to_list())

#print the number of terms (i.e., columns) in the document-term matrix
print(f"CountVectorizer - number of features: {len(vectorizer_CV_raw.get_feature_names_out())}")

#print the number of documents (i.e., rows) in the document-term matrix:
print(f"CountVectorizer - number of documents: {X_CV.shape[0]}")

#inspect the document-term matrix as a DataFrame for better readability - This is what the computer gets as input.
X_CV_df_raw = pd.DataFrame(X_CV.toarray(), columns=vectorizer_CV_raw.get_feature_names_out())
print("\nCountVectorizer - document-term matrix:\n")
print(X_CV_df_raw)

#Let's list the included features (i.e., words) in the vocabulary:
print("\nCountVectorizer - features (vocabulary):\n")
print(vectorizer_CV_raw.get_feature_names_out())

#Inspect the document-term matrix as a DataFrame for better readability - This is what the computer gets as input.
X_CV_df_raw = pd.DataFrame(X_CV.toarray(), columns=vectorizer_CV_raw.get_feature_names_out()) 
# The row numbers represent the sentences, the column labels represent the words in the vocabulary. 
# The values in the matrix represent the frequency of the word in the sentence.
# Note: We use sentences here, but the same applies to documents in general, including complete newspapers if you want to.
print("\nCountVectorizer - document-term matrix:\n")
print(X_CV_df_raw)
````

Take a good look at the document-term matrix:

Do you note how many words only occur in one document/sentence?

Do you note that each cell contains the **count** (frequency) of how many times that word appears in that document/sentence? (Here, very often 1)

This count is an integer (i.e., 0, 1, 2, etc.)

-> This is very different for TF-IDF, where features are weighted for uniqueness (and normalized), so they are a decimal number between 0 and 1

Also note that each document only contains a small set of words from the vocabulary

-> In contextual embeddings this is even more different, in that we represent documents in dimensions that no longer represent specific words. These dimensions do not tend to be either zero or one, so they make much more efficient use of the information that a computer can learn per cell (i.e., the value of a feature for a document)

````python
#spacy download:
import re #for regular expressions
import nltk #for nlp (here: tokenization)
from nltk.corpus import stopwords #for stopwords
from nltk.stem import WordNetLemmatizer #for lemmatization

def preprocess_text(texts, custom_stopwords=None):
    #NOTE: better lemmatizer that uses the WordNet lexical database.
    #NOTE: you can even pass it POS tags to improve the lemmatization (not done here).
    lemmatizer = WordNetLemmatizer() 
    #get the default stopwords from NLTK (en).
    stop_words = set(stopwords.words('english'))
    
    #add custom stopwords if provided
    if custom_stopwords:
        stop_words.update(custom_stopwords)
    
    #instantiate an empty list for processed texts
    processed_texts = []

    #interate over the texts
    for text in texts:
        try:
            #lowercase
            text = text.lower()

            #tokenize with NLTK tokenizer
            tokens = nltk.word_tokenize(text)
            
            #remove non-alphabetic characters from tokens
            tokens = [re.sub(r'[^a-z\s]', '', token) for token in tokens]
            
            #combined stopwords removal and lemmatization
            tokens = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]
            
            #join tokens back to string=
            processed_text = ' '.join(tokens)

            #remove any double spaces that remain:
            processed_text = re.sub(r'\s+', ' ', processed_text).strip()
            
            #append processed text to output list:
            processed_texts.append(processed_text)
        
        except Exception as e: #raise error if smth goes wrong: 
            print(f"Error processing text: {text}. Error: {e}")
            processed_texts.append("")
    
    return processed_texts
````

````python
#inspect the first 5 sentences
print(figure_df['Sentence'].to_list()[:5])

#execute the function:
processed_sentences = preprocess_text(figure_df['Sentence'].to_list())
print(processed_sentences[:5]) #inspect the first 5 processed sentences - what has changed?
````
